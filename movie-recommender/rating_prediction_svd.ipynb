# Collaborative filtering: rating prediction with SVD

```python
import pandas
```

```python
ratings = pandas.read_csv('ratings.csv')
ratings.head()
```

Create a Surprise dataset

```python
from surprise import Dataset , Reader
```

```python
reader = Reader(rating_scale=(1, 5))
dateset = Dataset.load_from_df(ratings[['userId', 'movieId', 'rating']], reader)
```

create trainset

```python
trainset = dateset.build_full_trainset()
```

TRAIN THE MODEL

```python
from surprise import SVD
```

```python
svd = SVD()
```

```python
svd.fit(trainset)
```

```python
svd.predict(15, 1956)
```

### VALIDATION

```python
from surprise.model_selection import cross_validate

```

```python
cross_validate(svd, dateset, measures=['RMSE', 'MAE'], cv=5, verbose=True)
```
